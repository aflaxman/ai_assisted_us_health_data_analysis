# NHANES Pulse Consumption Analysis

Replication and extension of Drewnowski et al. (2025) "Temporal trends in pulse
consumption in the US from 1999 to 2018" (*Frontiers in Nutrition*,
10.3389/fnut.2025.1638519).

## Notebooks

| # | Notebook | Description |
|---|---------|-------------|
| 01 | [Data Download](01_data_download.ipynb) | Downloads NHANES dietary recall + FPED + demographics for 10 cycles |
| 02 | [Analysis](02_analysis.ipynb) | Replicates abstract results, Table 1, and demographic breakdowns |

## Data sources

- **NHANES 1999–2018**: Individual Foods Day 1 (DR1IFF/DRXIFF) + Demographics (DEMO)
- **FPED/MPED**: USDA Food Patterns Equivalents Database for legume cup equivalents
- **DRXFCD/DRXFMT**: NHANES food code description files

## Key findings — replication comparison

| Metric | Our result | Paper |
|--------|-----------|-------|
| Total N | 48,736 | 48,738 |
| Consumer prevalence (weighted) | 18.9% | 17.2% |
| Mean oz eq/day (all) | 0.47 | 0.39 |
| Mean oz eq/day (consumers) | 2.46 | 2.26 |
| Mexican American prevalence | 35.3% | 33.3% |
| Non-Hispanic Black prevalence | 13.9% | 12.0% |

All key directional findings replicate:
- Mexican Americans have highest pulse consumption prevalence
- Non-Hispanic Blacks have lowest prevalence
- Men consume more than women
- Lower education and income associated with higher consumption
- Beans dominate; lentils and chickpeas associated with higher SES

The ~10% overestimation in our results likely stems from including
slightly more food codes as pulse-containing (475 vs paper's 339),
and using MPED 2003–2004 as a proxy for the 1999–2002 cycles.
