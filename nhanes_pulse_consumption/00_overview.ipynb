# NHANES Pulse Consumption Analysis

Replication and extension of Drewnowski et al. (2025) "Temporal trends in pulse
consumption in the US from 1999 to 2018" (*Frontiers in Nutrition*,
10.3389/fnut.2025.1638519).

## Notebooks

| # | Notebook | Description |
|---|---------|-------------|
| 01 | [Data Download](01_data_download.ipynb) | Downloads NHANES dietary recall + FPED + demographics for 10 cycles; builds broad and restricted pulse measures; adds nativity and origin variables |
| 02 | [Replication Analysis](02_analysis.ipynb) | Replicates abstract results, Table 1, temporal trends; sensitivity analysis comparing broad vs restricted food code measures; validation tests |
| 03 | [Nativity × Race/Ethnicity](03_country_of_origin.ipynb) | Pulse consumption by US-born vs foreign-born within Hispanic, NH Asian, NH White, NH Black, and Other/Multi-Racial groups; education stratification |

## Data sources

- **NHANES 1999–2018**: Individual Foods Day 1 (DR1IFF/DRXIFF) + Demographics (DEMO)
- **FPED/MPED**: USDA Food Patterns Equivalents Database for legume cup equivalents
- **DRXFCD/DRXFMT**: NHANES food code description files

## Key findings — replication comparison

**Broad measure** (all FPED-linked pulse codes, ~475 unique food codes):

| Metric | Broad | Restricted | Paper |
|--------|-------|-----------|-------|
| Total N | 48,736 | 48,736 | 48,738 |
| Consumer prevalence (weighted) | 18.9% | 18.1% | 17.2% |
| Mean oz eq/day (all) | 0.47 | 0.46 | 0.39 |
| Mean oz eq/day (consumers) | 2.46 | 2.57 | 2.26 |
| Mexican American prevalence | 35.3% | — | 33.3% |
| Non-Hispanic Black prevalence | 13.9% | — | 12.0% |

Restricting to each cycle's own FPED codes (V_LEGUMES ≥ 0.05) narrows the
prevalence gap but mean consumption still overestimates by ~18%, suggesting
additional classification differences beyond food code count.

## Key findings — nativity × race/ethnicity extension

Foreign-born individuals have higher pulse consumption across all race/ethnicity groups:

| Race/Ethnicity | US-born prev | Foreign-born prev | US-born oz/day | Foreign-born oz/day |
|---|---:|---:|---:|---:|
| Hispanic | 25.0% | 38.4% | 0.70 | 1.27 |
| NH Asian (2011+ only) | 20.1% | 29.3% | 0.27 | 0.73 |
| NH White | 16.7% | 18.1% | 0.36 | 0.42 |
| NH Black | 13.2% | 20.7% | 0.34 | 0.56 |

The nativity gap is largest for Hispanics and smallest for NH Whites.
Within each group, lower education is associated with higher consumption,
and this education gradient persists for both US-born and foreign-born.

## Limitations

- **South Asian vs East/SE Asian**: NHANES public data cannot distinguish these groups
- **Eastern vs Western European**: NHANES public data cannot distinguish these groups
- **NH Asian and Other/Multi-Racial**: Only available from 2011–2018 (RIDRETH3)
