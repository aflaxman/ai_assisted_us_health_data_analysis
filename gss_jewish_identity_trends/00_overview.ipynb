# GSS Jewish Identity Trends, 1972–2024

How well can the General Social Survey describe the demographics of American
Jews — and in particular, what can it say about the **"Jewish by religion"
vs. "Jews of no religion"** distinction that Pew's purpose-built surveys
measure directly? This project downloads the GSS 1972–2024 cumulative file,
tallies the trends, and shows how small the samples are and how noisy the
resulting estimates get.

**Data source:** GSS cumulative cross-sectional file (Release 3a,
`gss7224_r3a.dta`), NORC, <https://gss.norc.org/us/en/gss/get-the-data.html>

## Key results

**Sample sizes are tiny.** A GSS round interviews 1,400–4,500 adults, of whom
only **21–78 are Jewish by religion** (median 39 per round). Adults raised
Jewish number 22–86 per round. This is the fundamental limit on what the GSS
can say about American Jewish demographics.

**The estimates are correspondingly noisy.** Single-year 95% CIs span roughly
±1 percentage point on a ~2% share — half the estimate. The lowess trends
(notebook 02, figure 1) are far more informative than any single year.

| period | Jewish by religion | Jewish by religion or upbringing | raised Jewish, now no religion |
|--------|-------------------|----------------------------------|-------------------------------|
| 1970s | 2.3% | 2.6% | 9.4% |
| 1980s | 2.1% | 2.4% | 9.2% |
| 1990s | 2.0% | 2.3% | 8.6% |
| 2000s | 1.9% | 2.3% | 13.0% |
| 2010s | 1.9% | 2.2% | 11.8% |
| 2020s | 1.6% | 1.9% | 12.4% |

(Weighted, pooled within decade; see notebook 02 for yearly detail and CIs.)

**The "Jews of no religion" issue.** The GSS asks religion (`relig`) and
religion raised in (`relig16`) but never asks whether a respondent considers
themself Jewish *aside from* religion. So Pew-style "Jews of no religion" are
invisible in `relig` (they answer "none"). The closest proxy — adults raised
Jewish who now report no religion — has drifted from ~9% toward ~12%,
consistent in direction with Pew's finding that ~27% of Jewish Americans are
Jews of no religion, but not comparable in level: the GSS proxy misses people
raised outside Judaism who identify as Jewish, and includes people who no
longer identify as Jewish at all.

**Against purpose-built sources (notebook 03).** The GSS lowess trend for
Jewish by religion tracks the published by-religion estimates closely: NSRI
1990 and NJPS 1990/2000 (1.8%), Pew Religious Landscape Studies (1.7-1.9%),
Pew Jewish Americans 2020 (1.7%). ARIS 2001/2008 (1.3%, 1.2%) sit below every
other source, GSS included -- its open-ended question wording, not sampling
error, is the likely cause. The definitional split is bigger than most
between-source differences: Pew's net-Jewish estimates (2.2-2.4% of adults)
run ~0.6pp above by-religion ones. And a single Pew Jewish Americans survey
holds ~4,700 Jewish respondents -- three times as many as all 34 GSS rounds
combined (1,468).

**Age patterns (notebook 04).** The crude decline is a cohort/aging story,
not noise. Age-specific shares fell only under 65 (18-34: 1.8% -> 1.3%;
50-64: 3.1% -> 1.7% across four pooled eras) while the 65+ share rose
(2.6% -> ~3%). The Jewish-by-religion population aged far faster than
adults overall (65+ share: 18% -> 34%, vs 15% -> 21%), and the pooled
birth-cohort gradient declines monotonically from the 1910s cohort (3.2%)
to the 1970s cohort (1.3%). The 1990s-cohort uptick (1.8%, n=30) is the one
genuinely uncertain point.

**Knowledge, attitudes, practices (notebook 05).** Intermarriage among
married Jews-by-religion jumped from 15-20% (1972-98) to ~37% in 2012-24.
Jews sit far below Protestants and Catholics on every traditional
religiosity measure (weekly attendance 12% vs 37-38%; afterlife belief 45%,
below even the unaffiliated) yet match them on strong affiliation (42%) --
belonging without believing. Democratic identification eroded from 74% to
61% across the four eras. Non-Jewish respondents' feeling thermometer
toward Jews held steady around 60 (1986-2004, when the items stopped). Jews
average 7.2/10 on the vocabulary test vs 5.9 for Christians. And in
2008/2012 the GSS briefly asked the Pew-style "consider yourself Jewish"
screener: 19 of the 21 yes/part answers came from people NOT raised Jewish
-- confirming both failure modes of the upbringing-based proxy.

## Caveats

- `relig16` was first asked in 1973, so the raised-Jewish series starts there.
- The 2021+ rounds switched from in-person to mostly web administration; NORC
  cautions that this mode change can shift measured religiosity, so treat the
  recent downtick with extra care.
- Weighted with `wtssps` (defined for all rounds); CIs use the Kish effective
  sample size, which accounts for unequal weights but not the full complex
  survey design (no PSU/stratum clustering), so they are, if anything, a bit
  narrow.
